```python
import pandas as pd

# dataset from: https://huggingface.co/datasets/KisanVaani/agriculture-qa-english-only
df = pd.read_parquet("hf://datasets/KisanVaani/agriculture-qa-english-only/data/train-00000-of-00001.parquet")
df.to_excel("../db/qa_data/agriQA.xlsx",index=False)
```

```python
import json
import pandas as pd

df = pd.read_excel("../db/qa_data/agriQA.xlsx")

result = []
for idx, row in df.iterrows():
    # question 또는 answer가 NaN이면 스킵
    if pd.isna(row["question"]) or pd.isna(row["answers"]):
        continue
    
    item = {
        "id": idx,
        "question": row["question"],
        "answer": row["answers"],
        "ref_doc": "",
    }
    result.append(item)

# id 순서 재정렬
for new_id, item in enumerate(result):
    item["id"] = new_id

with open("../db/qa_data/agriQA_merged.json", "w", encoding="utf-8") as f:
    json.dump(result, f, ensure_ascii=False, indent=2)

print(f"총 {len(result)}개 저장 완료")
```
